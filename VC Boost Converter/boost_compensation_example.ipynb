```python
import math
```

```python
vin_min = 6 #V
vin_typ = 12 #V
vin_max = 18 #V

vout = 24 #V
iout = 2 #A

fsw = 440000 #Hz
inductance = 6.8*10**-6 #H
cout = 200*10**-6 #F
shunt_resistance = 0.008 #ohms

g_comp = 0.142 #V/V
g_m = 2*10**-3 #mho

r_load = vout/iout
duty_cycle_max = 1-(vin_min/vout)
duty_cycle_typ = 1-(vin_typ/vout)

f_z_rhp = (r_load*(1-duty_cycle_max)**2)/(2*3.1416*inductance)
fxo = min(fsw/10, f_z_rhp/5)
print(f"Crossover Frequency is {round(fxo*10**-3, 2)} kHz")

r_comp = (2*3.1416*cout*shunt_resistance*vout**2*fxo)/(g_comp*g_m*vin_min)
print(f"The Compensation Resistance is {round(r_comp*10**-3, 2)} kOhms")
r_comp_actual = 11300

c_comp = math.sqrt((cout*r_load)/(4*3.1416*r_comp_actual**2*fxo))
print(f"The Compensation Capacitance is {round(c_comp*10**9, 2)} nF")
c_comp_actual = 22*10**-9

c_hf = (c_comp_actual*inductance)/((c_comp_actual*(1-duty_cycle_typ)**2*r_load*r_comp_actual)-inductance)
print(f"The High Frequency Capacitance is {round(c_hf*10**12, 2)} pF")
```
