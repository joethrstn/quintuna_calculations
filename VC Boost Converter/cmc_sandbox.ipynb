```python
import math
```

```python
boost = 50
fsw = 700*10**3
R1 = 100*10**3
Rb = 3480
G = Rb/(R1 + Rb)
tan_arg = 45 + (boost/2)
K = math.tan(3.1416*(tan_arg/180))
C2 = 1/(2*3.1416*fsw*G*K*R1)
C1 = C2*((K**2)-1)
R2 = K/(2*3.1416*fsw*C1)

print(f"C_hf is {round(C2*10**12, 2)} pF")
print(f"C_comp is {round(C1*10**12, 2)} pF")
print(f"R_comp is {round(R2*10**-3, 2)} kOhms")
```
