```python
# System Parameters
max_segment_voltage = 4.2*14
min_segment_voltage = 400/10
print(min_segment_voltage)

vf_led = 1.7
led_resistor = 470
led_current = ((5-vf_led)/led_resistor)*10**3

ltc_current = 16.5 + 10.5 #mA
opamp_current = 2*4*300*10**-3 #mA (aka 300uA per op amp)
mux_current = 2*2.5

load_current = (10**-3)*(ltc_current + opamp_current + mux_current + led_current)
# load_current = load_current - 0.006
print(load_current)

```

```python
# Hi Side Resistor Calculations
hi_side_resistance = 100
# hi_side_resistance = (min_segment_voltage-6)/load_current
print(hi_side_resistance)
v_drop_resistor = load_current*hi_side_resistance
p_dis_resistor = (load_current**2)*hi_side_resistance
print(v_drop_resistor)
print(p_dis_resistor)
```

```python
# BJT thermals
v_drop_bjt = max_segment_voltage-v_drop_resistor-5
p_dis_bjt = v_drop_bjt * load_current
print(p_dis_bjt)

thermal_resistance_bjt = 62.5 #C/W
t_rise_bjt = p_dis_bjt*thermal_resistance_bjt
t_ambient = 60
t_bjt = t_ambient + t_rise_bjt
print(t_bjt)
```
