```python
max_pack_voltage = 14*10*4.2
min_pack_voltage = 60

current_5V = 0.008

maximum_resistance = min_pack_voltage/current_5V

resistances = [8250, 24000, 8250, 8250]

actual_resistance = sum(resistances)
print(f"Maximum Resistance is {maximum_resistance*10**-3} kOhms. Actual Resistance is {actual_resistance*10**-3} kOhms.")

i_min = min_pack_voltage/actual_resistance
i_max = max_pack_voltage/actual_resistance
print(f"I min is {round(i_min*10**3, 2)} mA")
print(f"I max is {round(i_max*10**3, 2)} mA")

for resistance in resistances:
    p_dis = resistance*i_max**2
    print(f"The power dissapation in the {round(resistance*10**-3,2)} kOhm resistor is {round(p_dis, 2)} Watts. The maximum voltage is {round(resistance*i_max,2)} V.")
```

```python
# Zener Diode Calculations

Vz = 5
zener_p_dis = Vz*i_max

print(f"Zener maximum dissapation is {round(zener_p_dis*10**3, 2)} mW")
```
